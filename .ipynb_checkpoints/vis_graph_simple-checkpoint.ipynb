```python
# import module 
import json
import graphviz
from graphviz import Digraph, Graph


file_path = '/home/siyuan/Documents/UIC_Falls__10707/md/INPATIENT/SpanTrex_OP/SpanT_note_48_Progress Notes_2022-01-15.json' # Location of json file to be viewed
#Generated image will be stored in path (directory + filename)
directory = '/home/siyuan/Documents/UIC_Falls__10707'
filename = 'Graphnew_view'

# instantiating object 
dot = Digraph(name='splines', engine='neato',graph_attr={'splines': 'true', 'overlap': 'scale' },comment='Trex_Visualization', node_attr={'style':'filled', 'fillcolor':'#40e0d0'}) 



try:
    with open(file_path, 'r') as file:
        data = json.load(file)
except FileNotFoundError:
        raise FileNotFoundError(f"Error: The file '{file_path}' was not found.")
except json.JSONDecodeError:
    raise json.JSONDecodeError(f"Error: Invalid JSON format in '{file_path}'.")

labels = {}
for id, node in data['entities'].items():
    dot.node(id,node['text'])
    labels[int(id)]=node['text']

print(labels)

for edge_data in data['reduced_temprels']:
    dot.edge(str(edge_data[0]), str(edge_data[1]), label=edge_data[2])#,constraint='false')#type=edge_data[2])


try:

    dot.format = 'png'
    print(dot.source)
    dot.render(filename=filename,directory= directory, view = False) 


except (FileNotFoundError, json.JSONDecodeError) as e:
    print(e)
```
